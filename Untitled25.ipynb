```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Make plots look professional
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10,6)
```

```python
# Upload your CSV file
df = pd.read_csv("/content/index_2.csv")


df.head()
```

```python
# Convert date and datetime columns into correct format
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')

# Make sure amount is numeric
df['money'] = pd.to_numeric(df['money'], errors='coerce')

# Create extra columns for analysis
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month_name()
df['weekday'] = df['date'].dt.day_name()
df['hour'] = df['datetime'].dt.hour
```

```python
print("Total Revenue:", df['money'].sum())
print("Average Order Value:", df['money'].mean())
print("Number of Transactions:", df.shape[0])
print("Unique Coffee Types:", df['coffee_name'].nunique())
```

```python
coffee_sales = df.groupby("coffee_name")['money'].sum().sort_values(ascending=False)

sns.barplot(x=coffee_sales.values, y=coffee_sales.index, palette="viridis")
plt.title("Revenue by Coffee Type")
plt.xlabel("Revenue")
plt.ylabel("Coffee")
plt.show()
```

```python
daily_sales = df.groupby("date")['money'].sum().reset_index()

sns.lineplot(x="date", y="money", data=daily_sales, marker="o")
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.histplot(df['hour'], bins=24, color="orange")
plt.title("Hourly Sales Distribution")
plt.xlabel("Hour of Day")
plt.ylabel("Number of Orders")
plt.show()
```

```python
weekday_sales = df.groupby("weekday")['money'].sum().reindex(
    ["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"]
)

sns.barplot(x=weekday_sales.index, y=weekday_sales.values, palette="coolwarm")
plt.title("Revenue by Weekday")
plt.xlabel("Weekday")
plt.ylabel("Revenue")
plt.show()
```

```python
payment_sales = df.groupby("cash_type")['money'].sum()

plt.pie(payment_sales.values, labels=payment_sales.index, autopct="%1.1f%%", startangle=90)
plt.title("Sales by Payment Type")
plt.show()
```

```python
heatmap_data = df.groupby(["weekday","hour"])['money'].sum().unstack(fill_value=0)

sns.heatmap(heatmap_data, cmap="YlGnBu")
plt.title("Sales Heatmap (Weekday vs Hour)")
plt.xlabel("Hour")
plt.ylabel("Weekday")
plt.show()
```
